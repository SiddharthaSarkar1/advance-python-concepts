```python
# Operator overloading

class Vector:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __add__(self, other):
    return Vector(self.x + other.x, self.y + other.y)

  def __repr__(self):
    return f"X: {self.x}, Y: {self.y}"

  def __len__(self):
    return 10 #hard codded a value for example purpose

  def __call__(self):
    print("Hello! I was called.")
```

```python
v1 = Vector(10, 20)
v2 = Vector(40, 50)
v3 = v1 + v2
```

```python
print(v3.x, v3.y)
```

```python
print(v3)
```

```python
print(len(v3))
```

```python
v3()
```

```python
# dunder methodes are methods with double underscores like __add__, also called as "magic methods"

class Microwave:
    def __init__(self, brand: str, power_rating: str) -> None:
        self.brand = brand
        self.power_rating = power_rating
        self.turned_on: bool = False

    def turn_on(self) -> None:
        if self.turned_on:
            print(f"Microwave {self.brand} is already turned on.")
        else:
            self.turned_on = True
            print(f"Microwave {self.brand} is now turned on.")

    def turn_off(self) -> None:
        if self.turned_on:
            self.turned_on = False
            print(f"Microwave {self.brand} is now turned off.")
        else:
            print(f"Microwave {self.brand} is already turned off.")

    def run(self, seconds: int) -> None:
        if self.turned_on:
            print(f"Microwave {self.brand} is running for {seconds} seconds.")
        else:
            print(f"Microwave {self.brand} is not turned on. Turn it on before running.")

    def __add__(self, other):
        return f"{self.brand} + {other.brand}"

    def __mul__(self, other):
        return f"{self.brand} * {other.brand}"

    def __str__(self) -> str:
        return f"Brand: {self.brand} (Power Rating: {self.power_rating})"

    def __repr__(self) -> str:
        return f"Microwave (brand = {self.brand}, power rating = '{self.power_rating}')"


smeg: Microwave = Microwave('Smeg', 'B')
bosch: Microwave = Microwave("Bosch", "C")

print(smeg + bosch)
print(smeg * bosch)

print(smeg)
print(bosch)

print(repr(smeg))
print(repr(bosch))
```

# Decorators

Decorators let you add extra behavior to a function, without changing the function's code.

A decorator is a function that takes another function as input and returns a new function.



```python
def myDecorator(function):
  def wrapper(*args, **kwargs):
    print("I am decorating your function.")
    function(*args, **kwargs)

  return wrapper
```

```python
@myDecorator
def hello_world():
  print("Hello World")
```

```python
hello_world()
```

```python
@myDecorator
def hello(person):
  print(f"Hello, {person}")
```

```python
hello("John")
```

```python
# Practical Example 1

def logged(function):
  def wrapper(*args, **kwargs):
    value = function(*args, **kwargs)
    with open("logfile.txt", "a+") as f:
      fname = function.__name__
      print(f"{fname} returned value as {value}")
      f.write(f"{fname} returned value as {value}\n")

    return value

  return wrapper
```

```python
@logged
def add(x, y):
  return x+y
```

```python
print(add(10,20))
```

```python
# Practical Example - 2

import time

def timed(function):
  def wrapper(*args, **kwargs):
    before = time.time()
    value = function(*args, **kwargs)
    after = time.time()
    fname = function.__name__
    print(f"{fname} took {after - before} seconds to execute.")
    return value

  return wrapper
```

```python
@timed
def myfunction(x):
  result = 1

  for i in range(1, x):
    result *= i
  return result
```

```python
myfunction(50)
```

```python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

def addgreeting(func):
  def myinner():
    return "Hello " + func() + " Have a good day!"
  return myinner

@changecase
@addgreeting
def myfunction():
  return "Tobias"

print(myfunction())
```

# Generators

```python
def mygenerator(n):
  for x in range(n):
    yield x ** 3
```

```python
values = mygenerator(100)

print(next(values)) # next statement always gives the next yield statement
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
```

```python
def my_generator():
  yield 1
  yield 2
  yield 3
  yield 4
  yield 5

for value in my_generator():
  print(value)
```

```python
# it will return value * 5
def infinite_sequence():
  result = 1
  while True:
    yield result
    result *= 5
```

```python
values = infinite_sequence()

print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
print(next(values))
```

# Argument Parsing

## *args

If you do not know how many arguments will be passed into your function, add a * before the parameter name.

This way, the function will receive a tuple of arguments and can access the items accordingly:

## *kwargs

If you do not know how many keyword arguments will be passed into your function, add two asterisks ** before the parameter name.

This way, the function will receive a dictionary of arguments and can access the items accordingly

```python
# Example of *args and **kwargs

def myfunction(*args, **kwargs):
  print(args[0])
  print(args[1])
  print(args[2])
  print(args[3])
  print(kwargs['KEYONE'])
  print(kwargs['KEYTWO'])

myfunction('Hey', True, 2026, 'wow', KEYONE="Sidd", KEYTWO=26)
```

# Encapsulation

```python
class Person:

  def __init__(self, name, age, gender) -> None:
     self.__name = name # __name, __age, __gender are private
     self.__age = age
     self.__gender = gender

  def get_name(self):
    return self.__name

  def set_name(self, value):
    self.__name = value


```

```python
p1 = Person("Darik", 21, "Male")

print(p1.get_name())
```

```python
p1.set_name("Derik Johnsom")
print(p1.get_name())
```

```python
# using @property and setter

class Personn:

  def __init__(self, name, age, gender) -> None:
     self.__name = name # __name, __age, __gender are private
     self.__age = age
     self.__gender = gender

  @property
  def Name(self):
    return self.__name

  @Name.setter
  def Name(self, value):
    self.__name = value

  @staticmethod
  def mystaticmethod():
    # static method can be called without object
    print("This is from the static method")
```

```python
p2 = Personn("Subrata", 27, "Male")
print(p2.Name)
```

```python
p2.mystaticmethod()
```

```python
Personn.mystaticmethod()
```

# Factory Design Pattern

```python
from abc import ABCMeta, abstractstaticmethod

class IMyPerson(metaclass=ABCMeta):
  # This is an abstract class, we cannot instantiate it

  @abstractstaticmethod
  def myperson_method():
    """ Interface Method """
```

```python
class Student(IMyPerson):
  """ Concrete Class """

  @staticmethod
  def myperson_method():
    print("I am a student")

  def student_method(self):
    print("I am a student method")

class Teacher(IMyPerson):
  """ Concrete Class """

  @staticmethod
  def myperson_method():
    print("I am a teacher")

  def teacher_method(self):
    print("I am a teacher method")
```

```python
s1 = Student()
s1.myperson_method()
```

```python
class MyPersonFactory:
  @staticmethod
  def build_person(person_type):
    if person_type == "Student":
      return Student()
    if person_type == "Teacher":
      return Teacher()

    print("Invalid Type")
    return -1

```

```python
choice = input("Enter the type of person you want to create?")
person = MyPersonFactory.build_person(choice)
person.myperson_method()
```

# Proxy Design Pattern


```python
class NewPerson(IMyPerson):
  @staticmethod
  def myperson_method():
    print("I am a Person")

  def new_person_method(self):
    print("I am a new person method")
```

```python
class ProxyPerson(IMyPerson):
  def __init__(self) -> None:
    self.person = NewPerson()

  @staticmethod
  def myperson_method():
    print("I am a Proxy Person")

  def new_person_method(self):
    print("I am the proxy functionality!")
    self.person.new_person_method()
```

```python
newp1 = NewPerson()
newp1.new_person_method()
```

```python
newp2 = ProxyPerson()
newp2.new_person_method()
```
